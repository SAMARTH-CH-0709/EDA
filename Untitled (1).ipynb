```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(42)
X = rng.random((10, 2))
print("Generated Points:")
print(X)
print("\nShape of X:")
print(X.shape)
plt.figure(figsize=(6, 5))
plt.scatter(
    X[:, 0],
    X[:, 1],
    s=80,
    edgecolors="black"
)
for i, point in enumerate(X):
    plt.annotate(
        str(i),
        point,
        textcoords="offset points",
        xytext=(0, 7),
        ha="center"
    )
plt.title("Initial Point Distribution (N = 10)")
plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
delta = X[:, None, :] - X[None, :, :]
dist_sq = np.einsum(
    "ijk,ijk->ij",
    delta,
    delta
)
print("\nPairwise Squared Distance Matrix:")
print(dist_sq)
print("\nShape of distance matrix:")
print(dist_sq.shape)
diagonal = np.diag(dist_sq)
print("\nDiagonal values:")
print(diagonal)
print("\nAre all diagonal values zero?")
print(np.allclose(diagonal, 0))
print("\nIs the distance matrix symmetric?")
print(np.allclose(dist_sq, dist_sq.T))
K = 2
sorted_indices = np.argsort(
    dist_sq,
    axis=1
)
print(sorted_indices)
print("first column values")
first_column = sorted_indices[:, 0]
print(first_column)
nearest_neighbours = sorted_indices[:, 1:K + 1]
print("\nTwo nearest neighbours of each point:")
print(nearest_neighbours)
candidate_indices = np.argpartition(
    dist_sq,
    kth=K,
    axis=1
)[:, :K + 1]
candidate_distances = np.take_along_axis(
    dist_sq,
    candidate_indices,
    axis=1
)
local_order = np.argsort(
    candidate_distances,
    axis=1
)
candidate_indices = np.take_along_axis(
    candidate_indices,
    local_order,
    axis=1
)
nearest_partition = candidate_indices[:, 1:K + 1]
print("\nNearest neighbours using argpartition:")
print(nearest_partition)
print("candidate_indices")
print(candidate_indices)
plt.figure(figsize=(7, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    s=90,
    edgecolors="black"
)
for i, point in enumerate(X):
    plt.annotate(
        str(i),
        point,
        textcoords="offset points",
        xytext=(0, 8),
        ha="center"
    )

for i in range(len(X)):
    for j in nearest_neighbours[i]:
        plt.plot(
            [X[i, 0], X[j, 0]],
            [X[i, 1], X[j, 1]]
        )
plt.title("Two-Nearest-Neighbour Network")
plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
# --------------------------------------------------
# 9. Performance Comparison - 6(a)
# --------------------------------------------------

def loop_neighbors(X):
    n = X.shape[0]
    neighbors = []
    for i in range(n):
        dist = ((X - X[i]) ** 2).sum(axis=1)
        order = np.argsort(dist)
        neighbors.append(order)
    return np.array(neighbors)
def vectorized_neighbors(X):
    dist = ((X[:, np.newaxis, :] -
             X[np.newaxis, :, :]) ** 2).sum(axis=2)
    order = np.argsort(dist, axis=1)
    return order
loop_result = loop_neighbors(X)
vectorized_result = vectorized_neighbors(X)
print("\nAre both results same?")
print(np.array_equal(loop_result, vectorized_result))

```

```python
N = 100
X_test = np.random.default_rng(42).random((N, 2))

print("N =", N)

print("Loop-based:")
%timeit loop_neighbors(X_test)

print("Vectorized:")
%timeit vectorized_neighbors(X_test)
```

```python
from sklearn.neighbors import KDTree
import numpy as np
pts = np.random.default_rng(0).random((1000, 2))
def kdtree_knn(points, k=2):
    tree = KDTree(points)
    _, idx = tree.query(points, k=k + 1)
    return idx[:, 1:]
numpy_result = vectorized_neighbors(pts)
kdtree_result = kdtree_knn(pts)
print("KDTree result shape:")
print(kdtree_result.shape)
print("\nNumPy result shape:")
print(numpy_result[:, 1:3].shape)
print("\nBoth methods find 2 nearest neighbours:")
print(kdtree_result.shape == numpy_result[:, 1:3].shape)
```
